### 1. Monitoring and Alerting in LangSmith

- Definition:
    - Monitoring in LangSmith looks across many traces at once to rack the overall health of your LLm system.

    - It aggregates key metrics like latency(P50, P95, P99) token usages, cost, error rates, and success rate. you can set up alerts to notify you when these metrics drift outside acceptable rages (eg: a spike in latency, higher error rates, or unexpected cost growth).


- Why it matter:
    - In production, issues often appear first as patterns across multiple runs rather than in a single trace.

    - Monitoring helps you catch these early signals before they impact users at scale, instead of waiting for customer complaints, you're proactively alerted when performance degrades or cost spike, enabling faster response and more reliable applications.
    
     
- Key Rule of Thump:
    - **Observability** is the study of single trace.
    - **Monitoring** are happening multiple traces and study together.
    - We can raise an **alert** based on specific metrics  

### 2. Evaluation in LangSmith

- What it does:
    - Evaluation in LangSmith helps you systematically measure the quality of your LLM outputs.

    - we can run tests against gold-standard datasets or apply custom evaluation metrics such as faithfulness, relevance, ro completeness.
    
    - LangSmith supports multiple approaches:
        - automated scoring with LLM-as-a-jude.
        - semantic similarity checks
        - even custom python evaluators.

    -  Evaluations can be run both offline(batch tests before deployment) and online(continuous checks on live traffic).


- What it matters:
    - LLM behavior can be unpredictable__ a small change in prompts, models, or retrieval logic may improve some cases but break others.

    - Evaluation provides an objective, repeatable way to track performance over time, ensuring that new versions are actually better and preventing regressions.

### 3. Prompt Experimentation

- What it does:
    - Prompt Experimentation in LangSmith allows you to systematically test and compare different prompt versions. 

    - you can run A/B tests across prompts on the same dataset, track their performance against evaluation metrics, and record the outcomes, Results are stored over time, giving you a clear history of which prompt variations worked best and under what conditions.


### 4. Dataset Creation and Annotation

- What it does:
    - Provides tools to build datasets for evaluation and fine-tuning.
    - Supports manual annotation (eg: labeling whether an answer is correct)
    - Stores datasets version for reuse across project.

- Why it matters?
    - High-quality datasets are critical for evaluation and feedback loops.

- Example:
    - Customer support: Build a dataset of common questions + expected answers.
    - Use it to benchmark you RAG agent every time you change retrieval logic.


### 5. User Feedback Integration
- What it does:
    - Lets you capture thumbs up/down, rating or structured feedback from user in production.

    - Feedback is logged alongside traces: 
        - tied to the exact prompt, model and state

    - support bul analysis of what user like/dislike.

### 6. Collaboration
- What it does:
    - Team members can view, share and comment on traces, dataset and evaluations.
    - Provides a web UI where non-engineer can inspect and annotate runs.
    - Enables shared experiment dashboards.
