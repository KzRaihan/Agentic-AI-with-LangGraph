### 1. Introduction of Agentic AI

- Definition: 
    - Agentic AI refers to AI systems that can autonomously plane, decide and execute actions to achieve a define goal without continuous human intervention.

    - Agentic AI is the design of AI systems that can pursue goals through dynamic decision-making, action, observation, and adaptation, often using models, tools, state, memory, and external environments.

####  1. Autonomous 
- Definition:
    - Autonomy refers to that AI system's ability to **make decision and take actions** on it's **own** to achieve a given goal, without needing step-by-step **human instructions**.


- Key Purpose of Autonomy:
- ==============================================
    - 1. It's Proactive.

    - 2. Autonomy is implement into multiple facets
        - i).   Execution point of views.

        - ii).  Decision Making views.

        - iii). Tool usage.

    - 3. Autonomy can be controlled
        - a). Permission Scope:
            - Limit what tools or actions the agents can perform independently.

        - b). Human-in-the-Loop(HITL):
            - Insert checkpoints where human approval is required before continuing.

        - c). Override Controls:
            - Allow users to stop, pause, or change the agent's behavior at any time.

        - d). Guardrails/Policies:
            - Define hard rules or ethical boundaries the agent must follow.


    - 4. Autonomy can be dangerous:
        - i). The application autonomously sends out job offers incorrect salaries or terms.

        - ii). The application spending extra on linkedLin ads.

```python
""" 
- The 6 Levels of AI Autonomy
============================================
    - Not all agents are equally autonomous. 
    - Think of this like self-driving car levels:


Level 0 ─── NO AUTONOMY
│            Traditional software. Rules-based. No AI.
│            Example: if/else chatbot
│
Level 1 ─── ASSISTED AI
│            AI suggests, human decides.
│            Example: Grammarly, autocomplete
│
Level 2 ─── PARTIAL AUTONOMY
│            AI handles simple tasks, human handles complex ones.
│            Example: ChatGPT with code interpreter
│
Level 3 ─── CONDITIONAL AUTONOMY  ← Most current agents
│            AI handles most tasks, human oversees.
│            Example: LangGraph agent with HITL
│            ⚠️  Human-in-the-loop is CRITICAL here
│
Level 4 ─── HIGH AUTONOMY
│            AI handles almost everything, human intervenes rarely.
│            Example: Devin (AI software engineer)
│
Level 5 ─── FULL AUTONOMY
             AI operates independently without human oversight.
             Example: Does not fully exist yet ⚠️



"""
```

#### 2. Goal oriented

- Definition:
    - An AI agent is generally goal-oriented, meaning it operates toward achieving a defined objective.

- Being goal-oriented means that the AI system operates with a **persistent** objective in mind and continuously directs it's **actions to achieve that objective**, rather than just responding to isolated prompts.

- 1. Goals acts as a compass for Autonomy
    - We can set the Goal as independent.
    - eg. (we hiring a ML engineer)

- 2. Goals can come with constraints.
    - eg. (we hiring a ML engineer who belong to Bangladesh)

- 3. Goals are stored in core memory.

- 4. Goals can be altered.
    

- Key Rule of Thump:
    - Goal-oriented = the agent's actions are directed toward achieving an objective.

#### 3. Planning

- Definition:
    - Planning is the agent's ability to **break down a high-level goal** into a structured sequence of actions or sub-goals and **decide the best path** to achieve the desired outcome.

    - Planning is the process of determining what steps should be performed and in what order to accomplish a goal.

    - Goal ---> Plane ---> Execution

- Example: 

    - Step 1: Generating multiple candidate plans
        - Plan A: Post JD on Linkedin, Github jobs, AngelList.
        - Plan B: Use internal referrals and hiring agencies.

    - Step 2: Evaluate a Plane
        - Efficiency
            - Which is faster?

        - Tools Availability
            - Which tools are available

        - Cost 
            - Does it required premium tools
        
        - Risk
            - Will it fail we get no applicants?
        
        - Alignment
            - With constrains (remote-only? budget?)

    - Step 3: Select the best plan with help of:
        
        - option 1: Human-in-the-loop integrate or input
            - eg: Which of these options do you prefer?

        - option 2: A Pre-Programmed policy
            - eg: Favor low-cost channels first


- Key idea
    - Planning = determining the sequence of actions needed to accomplish a goal.

```python
"""    - Example of Planning
     ===========================
Main Goal
│
├── Research
│   ├── Find sources
│   ├── Read sources
│   └── Extract findings
│
├── Analysis
│   ├── Compare findings
│   └── Identify patterns
│
└── Writing
    ├── Create outline
    ├── Write draft
    └── Review

- This is called task decomposition.
"""
```

#### 4. Reasoning
- Definition:
    - Reasoning is the cognitive process through which an agentic ai system interprets information, draws conclusions, and make decisions __ both while planning ahead and while executing actions in real time.


    - The agent's reasoning determines the appropriate action.

    - for each planning steps how to decide the execution are call Reasoning.

- 1. Reasoning During Planning:
    - a). Goal decomposition:
        - Break down abstract goals into concrete steps.

    - b). Tool selection:
        - Decide which tools will be needed for which steps.

    - c). Resource estimation:
        - Estimate time, dependencies tasks.


- 2. Reasoning During Execution:
    - a). Decision-making:
        - Choosing between options (eg: 3 candidates match --> schedule 2 best, reject 1)

    - b). HITL handling:
        - Knowing when to pause and ask for help (eg: Unsure about salary range)

    - c). Error Handling:
        - Interpreting tools/API failures and recovering.



- Key idea
    - Reasoning = using the available information and current state to determine an appropriate decision or next action.

```python
""" 
Example: Tool Selection
    - "What is 125 × 48?"

- The agent may determine: 

    Question
        ↓
    Requires calculation
        ↓
    Calculator tool is appropriate
        ↓
    Call calculator
        ↓
    Return result

"""
```

#### 5. Adaptability

- Definition:
    - Adaptability is the agent's ability to **modify it's plane, strategies, or actions** in response to unexpected conditions __ all while staying aligned with goal.


    - Adaptivity means the agent can change its behavior based on new information, feedback, or unexpected results.



- Key idea
    - Adaptivity = changing the next action based on feedback or changing circumstances.

```python
""" 
- Example: agent adapted to the situation:
==========================================

        Write code
            ↓
        Run tests
            ↓
        Test failed
            ↓
        Analyze error
            ↓
        Modify code
            ↓
        Run tests again
            ↓
        Test passed

"""
```

#### 6. Context Awareness
- Definition:
    - Context awareness is the agent's ability to **understand, retain, and utilize relevant information** from the ongoing task, past interactions, user , preferences, and environmental cues to make better decisions throughout a multi-step process.


    - Context awareness means an agent considers the current situation, available information, previous actions, user requirements, and relevant state when deciding what to do.


- 1. Types of Context an AI Agents can hold:
    - a). The original goal

    - b). Progress til now with interaction history

    - c). Environment state

    - d). Tools responses

    - e). User specific preferences

    - d). Policy or Guardrails

- 2. Context awareness is implemented through memory

- 3. Short term memory

- 4. Log term memory

### 3. Key Components of AI Agent

- The core high-level Component of AI Agents:
    - 1. Brain

    - 2. Orchestrator

    - 3. Tools

    - 4. Memory

    - 5. Supervisor


- ==========================================================

- 1. Brain:
    - Brain = intelligence/reasoning capability
    - Understands user instructions and translates them into object.
    - The Brain doesn't necessarily perform the external action itself.

- 2. Orchestrator:
    - Determines the order of actions (step 1 --> step2 -->....).
    - The Orchestrator controls the execution of the agent.
    - Orchestrator = controls how the agent operates.

- 3. Tools:
    - Tools give the agent capabilities that the model doesn't inherently have.
    - An LLM can generate a response, but it may need external tools to:
        - (Search the web, Query a database, Call an API, Calculate something, Read a file, Write a file, Execute code,Retrieve documents, Interact with external services)


- 4. Memory:
    - Maintains the active session's __ recent user messages, tools calls, and immediate decisions.
    - Memory allows an agentic system to retain and retrieve information beyond the immediate model call.
    - Memory = information the agent can retain and retrieve when needed.
         - (Short-term memory, Long-term memory, )


- 5. Supervisor:
    - The Supervisor is the governance or oversight layer.
    - Agents checks with humans before high risk actions (eg: sending offer letter)
    - Supervisor = governance, oversight, and intervention layer.

- Key Takeaways
    - An AI Agent is NOT just an LLM. It's an LLM embedded in a system with tools, memory, planning, and orchestration.

    - The agent loop is the defining feature. Think → Act → Observe → Repeat until done.


    - The LLM decides the control flow. This is the paradigm shift from traditional software.

    - There are multiple agent patterns. ReAct, Plan-and-Execute, Reflection, Multi-Agent — each for different use cases.


    - Autonomy is a spectrum. Most production agents are Level 2-3, not fully autonomous.


    - LangGraph is the orchestration layer. It turns an LLM into an agent by providing state management, routing, and persistence.
