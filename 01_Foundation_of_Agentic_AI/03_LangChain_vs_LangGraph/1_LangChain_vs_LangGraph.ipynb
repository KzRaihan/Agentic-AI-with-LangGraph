### 0. LangChain — The Building Blocks
- Definition:
    - LangChain is a framework for building applications powered by Large Language Models.

    - Think of LangChain as a toolkit and It provides building blocks:
        - LLMs
        - Prompt templates
        - Structured output
        - Embeddings
        - Vector databases
        - Retrievers
        - Tools
        - Document loaders
        - Output parsers
    
    - that we connect together in a linear sequence to solve tasks.

- Main idea: 
    - LangChain connects LLM components into a straight pipeline. Data flows in one direction — from input to output — like an assembly line.

    - [ LangChain = LLM + Prompts + Chains + Memory + Tools + Retrievers ]

```python
"""            LANGCHAIN ARCHITECTURE   
        --------------------------------
        
┌─────────────────────────────────────────────────────────────┐
│                  LANGCHAIN ARCHITECTURE                     │
│                                                             │
│  Components (Building Blocks):                              │
│                                                             │
│  ┌───────────────┐  ┌───────────────┐  ┌───────────────┐    │
│  │ Prompt        │  │ LLM / Chat    │  │ Output        │    │
│  │ Template      │  │ Model         │  │ Parser        │    │
│  └───────────────┘  └───────────────┘  └───────────────┘    │
│                                                             │
│  ┌───────────────┐  ┌───────────────┐  ┌───────────────┐    │
│  │ Document      │  │ Text          │  │ Embedding     │    │
│  │ Loader        │  │ Splitter      │  │ Model         │    │
│  └───────────────┘  └───────────────┘  └───────────────┘    │
│                                                             │
│  ┌───────────────┐  ┌───────────────┐  ┌───────────────┐    │
│  │ Vector Store  │  │ Retriever     │  │ Memory        │    │
│  │ (FAISS/Chroma)│  │               │  │ (Buffer/etc)  │    │
│  └───────────────┘  └───────────────┘  └───────────────┘    │
│                                                             │
│  How they connect:                                          │
│                                                             │
│  [Loader] → [Splitter] → [Embed] → [VectorStore]            │
│                                        ↓                    │
│  [Prompt] → [LLM] → [Parser]    [Retriever]                 │
│       ↑                              ↓                      │
│       └────── Context injected ──────┘                      │
│                                                             │
│  Pattern: LCEL (LangChain Expression Language)              │
│  chain = prompt | llm | parser                              │
│                                                             │
│  Data flows LEFT to RIGHT. No loops. No branches.           │
└─────────────────────────────────────────────────────────────┘

"""
```

### 1. Introduction of  LangGraph?
- Definition:
    - LangGraph is an orchestration framework that enables us to build stateful, multi-step, and event-driven workflow using large language model (LLMs).


    - LangGraph is a framework for building stateful, multi-actor, cyclic applications with LLMs. It extends LangChain by adding graph-based orchestration — loops, branches, conditional routing, and persistent state.


    - It's ideal for designing both single-agent and multi-agent in agentic AI Application.


- Think of langGraph as a flowchart engine for LLMs __ we define the steps(nodes), how they're connected (edges), and the logic that governs the transition. LangGraph takes core of state management conditional branching, looping, pausing/resuming and fault recovery features essential for building robust.


- Main idea: 
    - LangGraph lets you build loops and branches so the LLM can decide what to do next, retry on failure, and maintain state across steps.

    - LangGraph = LangChain + Graphs + Loops + State + Conditional Routing

```python
"""  
           LANGGRAPH ARCHITECTURE   
        --------------------------------
┌─────────────────────────────────────────────────────────────┐
│                  LANGGRAPH ARCHITECTURE                     │
│                                                             │
│  Core Concepts:                                             │
│                                                             │
│  1. STATE                                                   │
│     A shared data object that all nodes can read and write. │
│     Example: {"messages": [...], "plan": [...]}             │
│                                                             │
│  2. NODES                                                   │
│     Functions that receive state, do work, return updates.  │
│     Example: call_llm(state), run_tool(state)               │
│                                                             │
│  3. EDGES                                                   │
│     Connections between nodes.                              │
│     - Normal edge:      A → B (always)                      │
│     - Conditional edge: A → B or C (depends on state)       │
│                                                             │
│  4. GRAPH                                                   │
│     The complete network of nodes and edges.                │
│     Compiled into a runnable application.                   │
│                                                             │
│  Visual Structure:                                          │
│                                                             │
│         START                                               │
│           │                                                 │
│           ▼                                                 │
│      ┌─────────┐                                            │
│      │ Node A  │ ← Reads state, writes updates              │
│      └────┬────┘                                            │
│           │                                                 │
│      ┌────▼────┐                                            │
│      │ Router  │ ← Conditional edge (LLM decides)           │
│      └──┬───┬──┘                                            │
│     Yes │   │ No                                            │
│         ▼   ▼                                               │
│    ┌──────┐ ┌──────┐                                        │
│    │Node B│ │Node C│                                        │
│    │(Tool)│ │(End) │                                        │
│    └──┬───┘ └──────┘                                        │
│       │                                                     │
│       └──→ Back to Node A (LOOP!)                           │
│                                                             │
│  Pattern: StateGraph                                        │
│  graph = StateGraph(MyState)                                │
│  graph.add_node("agent", call_model)                        │
│  graph.add_node("tools", run_tools)                         │
│  graph.add_conditional_edges("agent", should_continue)      │
│  app = graph.compile(checkpointer=memory)                   │
│                                                             │
│  Data flows in ANY DIRECTION. Loops. Branches. Cycles.      │
└─────────────────────────────────────────────────────────────┘

"""
```

```python
"""     - Workflow Vs Agents
    =============================
- Workflow:
    - Workflow are system where LLMs and tools are orchestrated through predefine code path.
    - static Workflow that created by human.

- Agents:
    - Agents on the other hand, are system where LLM dynamically direct their own processes and tool usage, maintaining control over how they accomplish tasks.
    
    - Dynamic Workflow that created by Agents.


"""
```

### 2. Core Philosophy: Chain vs Graph

- LangChain → Chain: 
    - A chain traditionally represents a sequence:
    - [input ➔  Step A ➔ Step B ➔ Step C ➔ Step D ➔ Output]


- LangGraph → Graph:
    - A graph represents a network of states and transitions:

```python
""" - A graph can therefore represent complex control flow:
   
             ┌──────────────┐
             │    START     │
             └──────┬───────┘
                    ↓
             ┌──────────────┐
             │    Agent     │
             └──────┬───────┘
                    ↓
              Decision?
              /       \
            Yes        No
             │          │
             ▼          ▼
            Tool       Finish
             │
             ▼
          Agent
             │
             └──────────┐
                        │
                        ▼
                     Finish

"""
```

### 3. LangChain Vs LangGraph __ When to use What


- LangChain:
    - Use LangChain when you're building simple, linear workflows __ like a prompt chain, summarizer, or a basic retrieval system.

- LangGraph:
    - Use LangGraph when your use case involves complex, non-linear workflows that need:
        - 1. Conditional paths
        
        - 2. Loops
        
        - 3. Human-in-the-Loop steps

        - 4. Multi-agent coordination.

        - 5. Asynchronous or event-driven execution.


- Should We Still Use LangChain?
    - Yes, LangGraph is build on top of LangChain __ it doesn't replace it.

    - You'll still use LangChain components like:
        - 1. ChatOpenAl (LLMs)
        - 2. PromptTemplate
        - 3. Retrievers
        - 4. DocumentLoaders
        - 5. Tools and so on..

    - LangGraph handles Workflow orchestration, while LangChain provides the building blocks for each step in the workflow.


```python
""" 
            LANGCHAIN VS LANGGRAPH
        ---------------------------------


┌──────────────────────────────────────────────────────────────────┐
│                   LANGCHAIN VS LANGGRAPH                         │
│                                                                  │
│  DIMENSION            LANGCHAIN              LANGGRAPH           │
│  ──────────────────────────────────────────────────────────────  │
│  Structure            Linear chain           Cyclic graph        │
│                                                                  │
│  Control Flow         Fixed sequence         Dynamic routing     │
│                                                                  │
│  Loops                Not supported          Fully supported     │
│                                                                  │
│  Branches             Limited (if/else)      Conditional edges   │
│                                                                  │
│  State Management     Manual / limited       Built-in StateGraph │
│                                                                  │
│  Persistence          External needed        Built-in checkpoint │
│                                                                  │
│  Parallel Execution   Not native             Fan-out / Fan-in    │
│                                                                  │
│  Human-in-the-Loop    Difficult              Native support      │
│                                                                  │
│  Best For             Simple pipelines       Complex agents      │
│                                                                  │
│  Complexity           Low                    Medium              │
│                                                                  │
│  Learning Curve       Easy                   Moderate            │
│                                                                  │
│  Use Case Example     Q&A chain, RAG pipe    ReAct agent, CRAG   │
│                                                                  │
│  Analogy              Assembly line          Flowchart with      │
│                                            loops and branches    │
└──────────────────────────────────────────────────────────────────┘

"""
```

```python
""" 
        ->  When to Use LangChain
        --------------------------------

┌─────────────────────────────────────────────────────────────┐
│              USE LANGCHAIN WHEN...                          │
│                                                             │
│  ✅ Building a simple Q&A chain                             │
│     User asks → LLM answers. One step. Done.                │
│                                                             │
│  ✅ Building a basic RAG pipeline                           │
│     Retrieve → Prompt → Generate. Linear flow.              │
│                                                             │
│  ✅ Building a summarization pipeline                       │
│     Load document → Summarize → Output. No loops needed.    │
│                                                             │
│  ✅ Building a translation or formatting tool               │
│     Input text → Transform → Output. Single pass.           │
│                                                             │
│  ✅ Prototyping quickly                                     │
│     LCEL chains are fast to write and easy to understand.   │
│                                                             │
│  ✅ You do NOT need loops, retries, or dynamic routing      │
│                                                             │
│  EXAMPLE TASKS:                                             │
│  - "Summarize this PDF"                                     │
│  - "Translate this email to French"                         │
│  - "Answer questions about this document" (simple RAG)      │
│  - "Extract entities from this text"                        │
│  - "Convert this CSV to a report"                           │
└─────────────────────────────────────────────────────────────┘

"""
```

```python
""" 
        ->  When to Use LANGGRAPH
        --------------------------------

┌─────────────────────────────────────────────────────────────┐
│              USE LANGGRAPH WHEN...                          │
│                                                             │
│  ✅ Building an AI agent that uses tools                    │
│     The agent must decide WHICH tool to use and WHEN.       │
│     It may need to call multiple tools in sequence.         │
│                                                             │
│  ✅ Building a system that needs self-correction            │
│     "If the answer is wrong, try again." → Requires loop.   │
│                                                             │
│  ✅ Building a multi-step workflow with branches            │
│     "If relevant → generate. If not → search web."          │
│                                                             │
│  ✅ Building a system with persistent state                 │
│     Resume conversations, rewind to previous steps.         │
│                                                             │
│  ✅ Building Human-in-the-Loop systems                      │
│     Pause for approval, then continue.                      │
│                                                             │
│  ✅ Building multi-agent systems                            │
│     Researcher agent → Writer agent → Editor agent.         │
│                                                             │
│  ✅ Building Advanced RAG (CRAG, Self-RAG, Adaptive RAG)    │
│     These all require loops and conditional routing.        │
│                                                             │
│  EXAMPLE TASKS:                                             │
│  - "Research this topic and write a report" (multi-step)    │
│  - "Answer from docs, but verify with web search" (CRAG)    │
│  - "Process this refund, but ask manager if over $500"      │
│  - "Plan, research, write, review, and publish a blog"      │
│  - "Debug this code, run tests, fix failures, re-run"       │
└─────────────────────────────────────────────────────────────┘
        


"""
```

### 4. Why LangGraph exists?

```python
""" 
- The Complex Workflow of LangChain Challenges:
================================================

- 1. Challenge: Control Flow Complexity
    - 1. Conditional branching
    - 2. Loops
    - 3. Jump

    - LangChain are so suffering to do these tasks.

- 2. Challenge: Handling State (contain the importance of each state)
    - 1. LangChian is the Stateless

- 3. Challenge: Event Driven Execution/ PARALLEL EXECUTION 
    - LangChain sequential Execution (there is no Event driven execution) 

- 4. Challenge: Fault Tolerance
    - long running application can be Fault
    - small Fault
    - big Fault

    - LangChain has no Fault Tolerance.

- 5. Challenge: Human-in-the loop
    - LangChain HITL implement are difficult and are not useful.


- 6. Challenge: Nested Workflow
    - LangChain has no nested workflow or manually to do these.

- 7. Challenge: Observability
    - observability refers to how easily you can monitor, debug and understand what your workflow is doing at runtime.

    - LangSmith in LangChain, are only track the chain portion but can't track the others.
    - There are partial observability. 
    
""" 
```

#### Can They Work Together?
- Yes. LangGraph is built ON TOP of LangChain. They are not competitors — they are complementary.

```python
""" 
        ->  LANGCHAIN + LANGGRAPH TOGETHER 
        ------------------------------------

┌─────────────────────────────────────────────────────────────┐
│              LANGCHAIN + LANGGRAPH TOGETHER                 │
│                                                             │
│  LangGraph uses LangChain components INSIDE its nodes.      │
│                                                             │
│  ┌─────────────────────────────────────────────────────┐    │
│  │                  LANGGRAPH (Orchestrator)           │    │
│  │                                                     │    │
│  │  ┌──────────────┐  ┌──────────────┐  ┌───────────┐  │    │
│  │  │ Node: Agent  │  │ Node: RAG    │  │ Node:     │  │    │
│  │  │              │  │              │  │ Summarize │  │    │
│  │  │ Uses:        │  │ Uses:        │  │           │  │    │
│  │  │ • ChatOpenAI │  │ • Retriever  │  │ Uses:     │  │    │
│  │  │ • Tools      │  │ • Prompt     │  │ • LCEL    │  │    │
│  │  │ • Messages   │  │ • LLM Chain  │  │ • Chain   │  │    │
│  │  │              │  │              │  │           │  │    │
│  │  │ (LangChain)  │  │ (LangChain)  │  │(LangChain)│  │    │
│  │  └──────────────┘  └──────────────┘  └───────────┘  │    │
│  │                                                     │    │
│  └─────────────────────────────────────────────────────┘    │
│                                                             │
│  RULE OF THUMB:                                             │
│  ──────────────                                             │
│  • Use LangChain components (LLMs, prompts, retrievers,     │
│    parsers, tools) INSIDE your LangGraph nodes.             │
│                                                             │
│  • Use LangGraph to ORCHESTRATE the flow between nodes      │
│    (loops, branches, state, persistence).                   │
│                                                             │
│  LangChain = The bricks                                     │
│  LangGraph = The blueprint                                  │
└─────────────────────────────────────────────────────────────┘

"""
```

```python
"""
┌─────────────────────────────────────────────────────────────┐
│              WHY LANGGRAPH WAS CREATED                      │
│                                                             │
│  1. AGENTS NEED LOOPS                                       │
│     An agent tries a tool, checks the result, and tries     │
│     again if it failed. LangChain chains cannot loop.       │
│                                                             │
│  2. AGENTS NEED BRANCHES                                    │
│     "If the document is relevant, generate. If not,         │
│      search the web." LangChain chains go straight.         │
│                                                             │
│  3. AGENTS NEED STATE                                       │
│     The agent must remember what it has already done,       │
│     what tools it has called, and what results it got.      │
│                                                             │
│  4. AGENTS NEED PERSISTENCE                                 │
│     Pause a conversation, resume it tomorrow, rewind        │
│     to a previous step. Chains are stateless by default.    │
│                                                             │
│  5. AGENTS NEED HUMAN-IN-THE-LOOP                           │
│     Pause execution, wait for human approval, then          │
│     continue. Linear chains cannot pause and resume.        │
│                                                             │
│  6. AGENTS NEED PARALLEL EXECUTION                          │
│     Run 3 research tasks at the same time, then combine     │
│     results. Chains execute one step at a time.             │
└─────────────────────────────────────────────────────────────┘

"""
```

### 5. Key Takeaways

```python
""" 
        - Key Takeaways of LANGCHAIN VS LANGGRAPH 
        ------------------------------------------------

┌──────────────────────────────────────────────────────────────────┐
│                 LANGCHAIN VS LANGGRAPH                           │
│                                                                  │
│  LANGCHAIN:                                                      │
│  ──────────                                                      │
│  • Framework for building LLM-powered applications               │
│  • Linear chains: Prompt → LLM → Output                          │
│  • Components: Prompts, Models, Chains, Memory, Retrievers       │
│  • Best for: Simple pipelines, basic RAG, summarization          │
│  • Pattern: LCEL (chain = a | b | c)                             │
│  • Limitation: No loops, no dynamic routing, no state            │
│                                                                  │
│  LANGGRAPH:                                                      │
│  ──────────                                                      │
│  • Framework for building stateful, cyclic agent workflows       │
│  • Graph structure: Nodes + Edges + State + Conditions           │
│  • Components: StateGraph, Nodes, Edges, Checkpointer            │
│  • Best for: Agents, Advanced RAG, multi-step workflows          │
│  • Pattern: StateGraph (graph.add_node, add_conditional_edges)   │
│  • Advantage: Loops, branches, persistence, HITL, parallel       │
│                                                                  │
│  RELATIONSHIP:                                                   │
│  ─────────────                                                   │
│  • LangGraph is built ON TOP OF LangChain                        │
│  • LangChain provides the components (bricks)                    │
│  • LangGraph provides the orchestration (blueprint)              │
│  • Use BOTH together: LangChain inside LangGraph nodes           │
│                                                                  │
│  DECISION FRAMEWORK:                                             │
│  ───────────────────                                             │
│  "Can I solve this with a straight line?"                        │
│       YES → Use LangChain                                        │
│       NO  → Use LangGraph                                        │
│                                                                  │
│  "Does my task need loops, retries, or dynamic decisions?"       │
│       YES → Use LangGraph                                        │
│       NO  → Use LangChain                                        │
│                                                                  │
│  "Do I need to pause, resume, or remember across sessions?"      │
│       YES → Use LangGraph                                        │
│       NO  → Use LangChain                                        │
│                                                                  │
│  GOLDEN RULE:                                                    │
│  "Start with LangChain. When you hit a wall (need loops,         │
│   branches, or state), graduate to LangGraph."                   │
└──────────────────────────────────────────────────────────────────┘

- LangChain provides components for building LLM applications, while LangGraph provides the orchestration framework for building stateful, controllable, multi-step agentic workflows.

"""
```
