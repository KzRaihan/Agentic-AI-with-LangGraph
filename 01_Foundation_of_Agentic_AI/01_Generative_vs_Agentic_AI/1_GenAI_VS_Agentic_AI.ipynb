### 1. What is Generative AI?
- Definition:
    - Generative AI refers to a class of artificial intelligence models that can create new content __ such as text, image, audio, code or video that resembles human-created data.

=================================================================================================

- Generative AI Application:
    - 1. LLMs based app like Chat GPT
    
    - 2. Task Specific LLMs Models
        - for image data -> Diffusion model
        - for code generation -> CodeLLama model
        - for text-to-speech(TTS) generation -> ElevenLabs
        - for video generation -> Sora

    - 3. Creative and Business Writing.
    
    - 4. Software Development

    - 5. Customer Support

    - 6. Education Sector

    - 7. Designing


### 2. Generative AI Vs Traditional AI
- 
    - 1. Traditional AI
        - To find out the relation(pattern) between input and output data.
        - Traditional AI is about **finding patterns in data** and giving predictions.
        - Traditional AI is well-suite with Tabular Data.

        - example
            - Email spam detection
            - cancer detection
            - stock price prediction

    - 2. Generative AI
        - Generative AI is about **learning the distribution of data** so that it can generated a new sample from it.

        - where CV is well-suite of image data
        -  NLP is well-suite on sequence type or textual type of data.

        - The best part of Generative AI is that it can mimic humans.
        
        - Generative AI is constantly evolving and improving.

### 4. Generative AI Vs Agentic AI


- Generative AI is about creating content, Agentic AI is about solving a goal. 

- Generative AI is reactive, Agentic AI is Proactive(autonomous).

- Generative AI is a building block of Agentic AI

- Generative AI is a Capability, where as Agentic AI is a Beheviours.
