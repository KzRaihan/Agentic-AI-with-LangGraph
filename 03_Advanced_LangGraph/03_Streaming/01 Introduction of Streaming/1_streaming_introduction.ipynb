### Topic: Streaming Introduction

### 1. Introduction of Streaming

- Definition:
    - Streaming is a technique for transmitting and displaying data incrementally, piece-by-piece (in chunks), as soon as it becomes available—rather than waiting for the entire dataset or response to be fully processed/generated before showing anything to the user.



```python
""" 
NON-STREAMING (Blocking) Flow:
User Request ──> [Server processes ENTIRE response] ──> Full Response Sent ──> Displayed
                  |------------- Long Wait Time -------------|

STREAMING Flow:
User Request ──> Chunk 1 ──> Displayed Immediately
             ──> Chunk 2 ──> Displayed Immediately
             ──> Chunk 3 ──> Displayed Immediately
             ──> ... ──> Final Chunk ──> Complete


"""
```

### 2. Why Do We Use Streaming?

- 1. Perceived Performance & Reduced Latency

- 2. Better User Experience (UX)

- 3. Faster response time - low drop-off rates

- 4. Mimics human like conversation(builds trust, feels alive and keeps the user engaged)

- 5. Important for Multi-Modal UIs

- 6. we can cancel midway saving tokens

- 7. we can interleave UI upates, 
    - eg. show "thinking...." sho tools results.




- Real-Time Use Cases
- Essential for:
    - Live video/audio broadcasting
    - Live stock market tickers
    - Real-time sensor/IoT data feeds
    - AI Chatbots (LLM token streaming)

### How to use the streaming 

- Instead of using graph.invoke({}), we use the graph.stream({})

- The graph.stream({}) method return a generator object which contain two value 
    - first : message_chunk
    - second: metadata 

- What is generator:
    - In python, a generator is a special type of iterator that allows us to generate values on the fly, one at a time, using the yield keyword instead of return.

- the stream_mode = messages — Token-by-Token LLM Output 

- for streaming in streamlit web interface we use : 
    - st.write_stream(my_generator)

```python
# streaming 
stream_generator = chatbot.stream(
    {
        "message": [HumanMessage(content="What is the recipe to make biryani")]
    },
    CONFIG = {
    "configurable":{
        "thread_id": "1"
        }
    }, 
    stream_mode="messages"
)

# now display the generator object output  Token-by-Token 
# Generator return two values _ 1. message_chunk and 2. metadata

# we only display the message_chunk
for message_chunk, metedata in stream_generator:
    if message_chunk.content:
        print(message_chunk.content, end=" ", flush=True)
    
```

```python
# Final backend Streaming code after chatbot compile 
for message_chunk, metadata in chatbot.stream(
    {
        "messages": [
            HumanMessage(content="What is the recipe to make pasta")
        ]
    },
    config={
        "configurable": {
            "thread_id": "thread-1"
        }
    },
    stream_mode="messages"
):
    if message_chunk.content:
        print(message_chunk.content, end="", flush=True)
```
