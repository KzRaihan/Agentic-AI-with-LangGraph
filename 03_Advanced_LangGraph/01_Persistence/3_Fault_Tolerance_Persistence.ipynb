```python
# import necessary Library
from langgraph.graph import StateGraph, START, END
from langgraph.checkpoint.memory import InMemorySaver

from typing import TypedDict

import time
```

```python
# 1. Define the state
class CrashState(TypedDict):
  input: str
  step1: str
  step2: str
  step3: str
```

```python
# 2. Define steps
def step_1(state: CrashState) -> CrashState:
  print("Step 1 Exectued")

  return {
      "step1": "Done",
      "input": state['input']
  }

def step_2(state: CrashState) -> CrashState:
  print("Step 2 hanging.... now manually interrput from the notebook toolbar (STOP button)")

  time.sleep(30) # Simulate long-running hung

  return {
      "step2": "Done"
  }

def step_3(state: CrashState) -> CrashState:
  print("Step 3 Exectued")

  return {
      "step3": "Done"
  }
```

```python
# Define Graph
graph = StateGraph(CrashState)

# nodes
graph.add_node("step_1", step_1)
graph.add_node("step_2", step_2)
graph.add_node("step_3", step_3)

# edges
graph.add_edge(START, "step_1")
graph.add_edge("step_1", "step_2")
graph.add_edge("step_2", "step_3")
graph.add_edge("step_3", END)


# Define memory
memory = InMemorySaver()

# compile
workflow = graph.compile(checkpointer=memory)

workflow
```

```python
# execute the graph

try:
  workflow.invoke({
      "input": "start"
  }, config= {
      "configurable": {"thread_id": "thread-1"}
  })

except KeyboardInterrupt:
  print("Kernal manually interrupted (Crash simulated)")
```

- Key Note:
  - Here, we manually interrput or Crash simulate.

```python
# for crash simuate the final state value
workflow.get_state({
    "configurable": {"thread_id": "thread-1"}
})
```

- Key note:
  - here, only state 1 is executed but next is step 2 (this is not execute)

```python
# for crash simuate the all state values
list(workflow.get_state_history({
    "configurable": {"thread_id": "thread-1"}
}))
```

```python
# # Now Re-run the workflow and show fault-tolerant resume
try:
    workflow.invoke(
        {"input": "start"},
        config={
            "configurable": {
                "thread_id": "thread-1"
            }
        }
    )

except RuntimeError as e:
    print(f"Workflow crashed: {e}")
```

```python
state = workflow.get_state(
    {
        "configurable": {
            "thread_id": "thread-1"
        }
    }
)

print(state)
```
